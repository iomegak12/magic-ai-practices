```python
import os

from dotenv import load_dotenv
from agent_framework import Agent
from agent_framework.foundry import FoundryAgent
from azure.identity import AzureCliCredential
```

```python
load_dotenv(override=True)

project_endpoint = os.getenv("AZURE_AI_PROJECT_ENDPOINT")
model = os.getenv("AZURE_OPENAI_RESPONSES_DEPLOYMENT_NAME")

print("Project Endpoint: ", project_endpoint)
print("Model: ", model)
```

```python
agent = FoundryAgent(
    project_endpoint=project_endpoint,
    agent_name="ramkumar-rag-agent",
    agent_version="2",
    credential=AzureCliCredential(),
)
```

```python
session = agent.create_session()

query = "Hello! My name is Kritika. what are all the hotels offered by Margies Travels in Dubai?"

print(f"User: {query}")
print(f"Agent: {await agent.run(query, session=session)}\n")
```
