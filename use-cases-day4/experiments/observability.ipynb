```python
import asyncio
from random import randint
from typing import Annotated

from agent_framework import Agent, tool
from agent_framework.observability import configure_otel_providers, get_tracer
from opentelemetry.trace import SpanKind
from opentelemetry.trace.span import format_trace_id
from pydantic import Field
from agent_framework.openai import OpenAIChatClient
from azure.identity.aio import AzureCliCredential

import os
from dotenv import load_dotenv
```

```python
load_dotenv(override=True)

azure_endpoint = os.getenv("AZURE_OPENAI_ENDPOINT")
model = os.getenv("AZURE_OPENAI_RESPONSES_DEPLOYMENT_NAME")

print("Azure OpenAI Endpoint: ", azure_endpoint)
print("Model: ", model)
```

```python
@tool(approval_mode="never_require")
async def get_weather(
    location: Annotated[str, Field(description="The location to get the weather for.")],
) -> str:
    """Get the weather for a given location."""
    await asyncio.sleep(randint(0, 10) / 10.0)  # Simulate a network call
    conditions = ["sunny", "cloudy", "rainy", "stormy"]
    return f"The weather in {location} is {conditions[randint(0, 3)]} with a high of {randint(10, 30)}°C."
```

```python
configure_otel_providers()
```

```python
questions = ["What's the weather in Amsterdam?",
             "and in Paris, and which is better?", "Why is the sky blue?"]

with get_tracer().start_as_current_span("Scenario: Simple Agent Chat", kind=SpanKind.CLIENT) as current_span:
    print(
        f"Trace ID: {format_trace_id(current_span.get_span_context().trace_id)}")

    credential = AzureCliCredential()
    client = OpenAIChatClient(
        model=model,
        azure_endpoint=azure_endpoint,
        credential=credential,
    )

    agent = client.as_agent(
        name="SimpleAgent",
        instructions=(
            "You are a Simple Service Agent with access to one system:\n"
            "1. Weather Information - for looking up current weather conditions in various cities.\n\n"
            "When providing weather information, use context from previous conversation turns to select the "
            "appropriate city and provide relevant details. "
            "Always confirm the action taken and include relevant information in your response."
        ),
        tools=[get_weather]
    )

    thread = agent.create_session()

    for question in questions:
        print(f"\nUser: {question}")
        print(f"{agent.name}: ", end="")
        async for update in agent.run(
            question,
            session=thread,
            stream=True,
        ):
            if update.text:
                print(update.text, end="")
```
