```python
import os
from dotenv import load_dotenv
from agent_framework.azure import AzureOpenAIResponsesClient
from azure.identity.aio import AzureCliCredential
from agent_framework import MCPStreamableHTTPTool
```

```python
load_dotenv(override=True)

project_endpoint = os.getenv("AZURE_AI_PROJECT_ENDPOINT")
model = os.getenv("AZURE_OPENAI_RESPONSES_DEPLOYMENT_NAME")

print("Project Endpoint: ", project_endpoint)
print("Model: ", model)
```

```python
complaints_management_mcp_tool = MCPStreamableHTTPTool(
    name = "Complaints Management Server",
    url = "http://localhost:8000/mcp"
)
```

```python
credential = AzureCliCredential()
client = AzureOpenAIResponsesClient(
    project_endpoint=project_endpoint,
    deployment_name=model,
    credential=credential,
)
agent = client.as_agent(
    name="ComplaintsManagementAgent",
    instructions="You are a helpful assistant that can help with Complaints Management Related activities.",
    tools=[complaints_management_mcp_tool]
)
```

```python
session = agent.create_session()
```

```python
query = "My name is Ramkumar, i would like to register a complaint regarding the product i purchased last week. Can you please help me with that?"
response = await agent.run(query, session=session)
```

```python
print("\nAssistant:\n", response)
```
